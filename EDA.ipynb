## Consistency in expected values for the 6 attributes

````python
# ============================================================================
# CONSISTENCY ANALYSIS: Identify inconsistent samples
# ============================================================================
# For each sample, check ONLY the attributes that are present and SPECIFIED.
# "Not provided" values are IGNORED (not counted as mismatches).
# Uses MAJORITY VOTE: if >50% of specified values match expected, sample is consistent.

print("\n\n" + "=" * 120)
print("CONSISTENCY ANALYSIS: Majority Vote with 'Not Provided' Ignored")
print("=" * 120)

# Define gut attributes with acceptable values (case-insensitive matching)
gut_attributes = {
    'description': 'American Gut Project Stool Sample',
    'environmental package': 'human-gut',
    'environmental medium': 'feces',
    'body habitat': 'UBERON:feces',
    'body product': 'UBERON:feces',
    'sample type': ['Stool', 'feces']  # Accept both Stool and feces (case-insensitive)
}

def value_matches(actual_value, expected_value):
    """
    Check if actual value matches expected value.
    Handles both single values and lists.
    Case-insensitive matching.
    """
    if actual_value.lower() == 'not provided':
        return None  # None means "not specified, ignore"
    
    if isinstance(expected_value, list):
        # For lists, check if actual matches any item (case-insensitive)
        return any(actual_value.lower() == exp.lower() for exp in expected_value)
    else:
        # For single values, case-insensitive comparison
        return actual_value.lower() == expected_value.lower()

# Analyze each sample
perfect_100_samples = []     # 100% match
partial_50_to_100_samples = []  # >50% but <100% match
partial_match_samples = []   # Some match, some don't (but not majority)
complete_mismatch_samples = []  # None match or no specified values to evaluate

sample_details = {}  # Store details of all non-perfect samples

for sample_id in NCBI_dict.keys():
    attributes = NCBI_dict[sample_id]['Attributes']
    
    # Find which gut attributes are present in this sample
    present_attrs = {k: v for k, v in attributes.items() if k in gut_attributes}
    
    if not present_attrs:
        # Skip samples with none of the gut attributes
        continue
    
    # Check which present attributes match expected values
    # Ignore "Not provided" values
    matches = 0
    specified_count = 0  # Count of non-"not provided" attributes
    attr_results = {}
    
    for attr_name, actual_value in present_attrs.items():
        expected_value = gut_attributes[attr_name]
        match_result = value_matches(actual_value, expected_value)
        
        # Format expected value for display
        if isinstance(expected_value, list):
            expected_display = " or ".join([f"'{v}'" for v in expected_value])
        else:
            expected_display = expected_value
        
        attr_results[attr_name] = {
            'expected': expected_display,
            'actual': actual_value,
            'match': match_result  # True, False, or None (not specified)
        }
        
        if match_result is None:
            # "Not provided" - ignore it
            pass
        elif match_result:
            matches += 1
            specified_count += 1
        else:
            specified_count += 1
    
    # Calculate match percentage
    match_percentage = (matches / specified_count * 100) if specified_count > 0 else 0
    
    # Categorize sample using majority vote
    if specified_count == 0:
        # All attributes are "Not provided" - can't evaluate
        complete_mismatch_samples.append(sample_id)
        sample_details[sample_id] = attr_results
    elif match_percentage == 100:
        # Perfect match (100%) - PERFECT
        perfect_100_samples.append(sample_id)
    elif matches >= specified_count / 2:
        # Majority match (>50% but <100%) - CONSISTENT
        partial_50_to_100_samples.append(sample_id)
    elif matches == 0:
        # No matches at all - COMPLETE MISMATCH
        complete_mismatch_samples.append(sample_id)
        sample_details[sample_id] = attr_results
    else:
        # Some match, but not majority - PARTIAL
        partial_match_samples.append(sample_id)
        sample_details[sample_id] = attr_results

# Print statistics
print(f"\n[SUMMARY] Sample Consistency Breakdown (Case-Insensitive, Majority Vote)")
print("-" * 120)
total_samples = len(perfect_100_samples) + len(partial_50_to_100_samples) + len(partial_match_samples) + len(complete_mismatch_samples)
print(f"\nTotal samples with any gut attributes: {total_samples:,}")
print(f"\n{'Category':<50} {'# Samples':<20} {'% of Total':<20}")
print("-" * 120)
print(f"{'Perfect Match (100%)':<50} {len(perfect_100_samples):<20} {len(perfect_100_samples)/total_samples*100:<19.1f}%")
print(f"{'Consistent (>50% to <100%)':<50} {len(partial_50_to_100_samples):<20} {len(partial_50_to_100_samples)/total_samples*100:<19.1f}%")
print(f"{'TOTAL ≥50% MATCH (GUT SAMPLES)':<50} {len(perfect_100_samples) + len(partial_50_to_100_samples):<20} {(len(perfect_100_samples) + len(partial_50_to_100_samples))/total_samples*100:<19.1f}%")
print(f"{'Partial Match (<50%)':<50} {len(partial_match_samples):<20} {len(partial_match_samples)/total_samples*100:<19.1f}%")
print(f"{'Complete Mismatch (0% match)':<50} {len(complete_mismatch_samples):<20} {len(complete_mismatch_samples)/total_samples*100:<19.1f}%")

# ============================================================================
# DETAILED LOOK AT PARTIAL MATCH SAMPLES
# ============================================================================
print(f"\n\n[PARTIAL MATCHES] Samples with Some Matches (<50%)")
print("-" * 120)

print(f"\nTotal partial match samples: {len(partial_match_samples)}")
print(f"\nShowing first 30 samples:\n")

for idx, sample_id in enumerate(partial_match_samples[:30]):
    attr_results = sample_details[sample_id]
    
    # Count matches/mismatches/unspecified for this sample
    matches_count = sum(1 for r in attr_results.values() if r['match'] is True)
    mismatches_count = sum(1 for r in attr_results.values() if r['match'] is False)
    unspecified_count = sum(1 for r in attr_results.values() if r['match'] is None)
    
    print(f"\n[Sample {idx+1}] {sample_id}")
    print(f"  Status: {matches_count} match, {mismatches_count} mismatch, {unspecified_count} not provided")
    print(f"  ─" * 60)
    
    for attr_name in sorted(attr_results.keys()):
        result = attr_results[attr_name]
        if result['match'] is None:
            status = "⊘ NOT PROVIDED"
        elif result['match']:
            status = "✓ MATCH"
        else:
            status = "✗ MISMATCH"
        
        actual_display = result['actual'][:50] + "..." if len(result['actual']) > 50 else result['actual']
        
        print(f"  {attr_name:<35} {status}")
        print(f"    Expected: {result['expected']}")
        print(f"    Actual:   {actual_display}")

if len(partial_match_samples) > 30:
    print(f"\n... and {len(partial_match_samples) - 30} more partial match samples")

# ============================================================================
# COMPLETE MISMATCH ANALYSIS - First 20 samples
# ============================================================================
print(f"\n\n[COMPLETE MISMATCHES] Samples with No Matches")
print("-" * 120)

print(f"\nTotal complete mismatch samples: {len(complete_mismatch_samples):,}")
print(f"\nShowing first 5 samples:\n")

for idx, sample_id in enumerate(complete_mismatch_samples[:5]):
    attr_results = sample_details.get(sample_id, {})
    
    if not attr_results:
        print(f"\n[Sample {idx+1}] {sample_id}")
        print(f"  Status: No gut attributes to evaluate")
        continue
    
    # Count matches/mismatches/unspecified for this sample
    matches_count = sum(1 for r in attr_results.values() if r['match'] is True)
    mismatches_count = sum(1 for r in attr_results.values() if r['match'] is False)
    unspecified_count = sum(1 for r in attr_results.values() if r['match'] is None)
    
    print(f"\n[Sample {idx+1}] {sample_id}")
    print(f"  Status: {matches_count} match, {mismatches_count} mismatch, {unspecified_count} not provided")
    print(f"  ─" * 60)
    
    for attr_name in sorted(attr_results.keys()):
        result = attr_results[attr_name]
        if result['match'] is None:
            status = "⊘ NOT PROVIDED"
        elif result['match']:
            status = "✓ MATCH"
        else:
            status = "✗ MISMATCH"
        
        actual_display = result['actual'][:50] + "..." if len(result['actual']) > 50 else result['actual']
        
        print(f"  {attr_name:<35} {status}")
        print(f"    Expected: {result['expected']}")
        print(f"    Actual:   {actual_display}")

if len(complete_mismatch_samples) > 5:
    print(f"\n... and {len(complete_mismatch_samples) - 5} more complete mismatch samples")

# ============================================================================
# ANALYZE PATTERNS IN PARTIAL MATCHES
# ============================================================================
print(f"\n\n[PATTERNS] Common Attribute Patterns in Partial Matches")
print("-" * 120)

# For partial match samples, see which attributes typically match/mismatch
mismatch_frequency = {}
match_frequency = {}
unspecified_frequency = {}

for sample_id in partial_match_samples:
    attr_results = sample_details[sample_id]
    for attr_name, result in attr_results.items():
        if result['match'] is True:
            if attr_name not in match_frequency:
                match_frequency[attr_name] = 0
            match_frequency[attr_name] += 1
        elif result['match'] is False:
            if attr_name not in mismatch_frequency:
                mismatch_frequency[attr_name] = 0
            mismatch_frequency[attr_name] += 1
        else:  # None - not provided
            if attr_name not in unspecified_frequency:
                unspecified_frequency[attr_name] = 0
            unspecified_frequency[attr_name] += 1

print(f"\nAmong {len(partial_match_samples)} partial match samples:\n")
print(f"{'Attribute':<35} {'Matches':<15} {'Mismatches':<15} {'Not Provided':<15} {'Match %':<10}")
print("-" * 120)

for attr_name in sorted(gut_attributes.keys()):
    matches = match_frequency.get(attr_name, 0)
    mismatches = mismatch_frequency.get(attr_name, 0)
    unspecified = unspecified_frequency.get(attr_name, 0)
    total = matches + mismatches + unspecified
    
    if total > 0:
        match_pct = (matches / (matches + mismatches) * 100) if (matches + mismatches) > 0 else 0
        print(f"{attr_name:<35} {matches:<15} {mismatches:<15} {unspecified:<15} {match_pct:<9.1f}%")
````

The 14 samples classified as "inconsistent" were inspected separately and showed that the mismatched attributes had values such as "not provided" or "not collected", thus the >50% rule was implemented.

# Count "Not provided" values

````python
# Analyze "Not provided" values for all samples
print(f"Analyzing 'Not provided' values for all {len(gut_NCBI_dict)} samples...")

def count_not_provided_values(sample_id):
    """Count how many attributes have 'Not provided' values for a given sample"""
    if sample_id not in gut_NCBI_dict:
        return None, None
    
    attributes = gut_NCBI_dict[sample_id]['Attributes']
    total_attributes = len(attributes)
    not_provided_count = 0
    
    for attr_name, attr_value in attributes.items():
        if attr_value == 'Not provided':
            not_provided_count += 1
    
    return not_provided_count, total_attributes

# Store results for analysis
not_provided_results = {}
sample_stats = []

for sample_id in gut_NCBI_dict.keys():
    not_provided, total = count_not_provided_values(sample_id)
    if not_provided is not None:
        percentage = (not_provided / total) * 100
        not_provided_results[sample_id] = {
            'not_provided': not_provided,
            'total': total,
            'percentage': percentage
        }
        sample_stats.append({
            'sample_id': sample_id,
            'not_provided': not_provided,
            'total': total,
            'percentage': percentage
        })

# Convert to DataFrame for easier analysis
df_stats = pd.DataFrame(sample_stats)

print(f"\n=== Summary Statistics ===")
print(f"Total samples analyzed: {len(df_stats)}")
print(f"Average 'Not provided' per sample: {df_stats['not_provided'].mean():.1f}")
print(f"Average total attributes per sample: {df_stats['total'].mean():.1f}")
print(f"Average percentage 'Not provided': {df_stats['percentage'].mean():.1f}%")
print(f"Min percentage 'Not provided': {df_stats['percentage'].min():.1f}%")
print(f"Max percentage 'Not provided': {df_stats['percentage'].max():.1f}%")
````

## Missing and unique subset values

````python
# Check for NaN and unique values across ALL gut samples for subset attributes
print("=" * 100)
print("SUBSET ATTRIBUTES: Check for NaN/None values and unique values across all samples")
print("=" * 100)

subset_attrs = ['subset_age', 'subset_antibiotic_history', 'subset_bmi', 'subset_diabetes', 'subset_healthy', 'subset_ibd']

for attr_name in subset_attrs:
    values_list = []
    nan_count = 0
    
    # Collect all values for this attribute across all samples
    for sample_id in gut_NCBI_dict.keys():
        attrs = gut_NCBI_dict[sample_id]['Attributes']
        if attr_name in attrs:
            value = attrs[attr_name]
            # Check if value is NaN or None
            if value is None or (isinstance(value, float) and pd.isna(value)) or value == 'nan':
                nan_count += 1
            else:
                values_list.append(value) # if it's a valid value, add to list for unique value analysis
        else:
            nan_count += 1
    
    # Get unique values
    unique_vals = set(values_list) # use set to get unique values from the list of valid values
    
    print(f"\n{attr_name}:")
    print(f"  Total samples: {len(gut_NCBI_dict)}")
    print(f"  NaN/None values: {nan_count}")
    print(f"  Valid values: {len(values_list)}")
    print(f"  Unique values: {sorted(unique_vals)}")
````

## Proportion of True/False and missing subset attributes in the entire dataset

````python
# Dictionary to store counts for each attribute
subset_counts = {}

# Count true/false values for each subset attribute across all samples
for attr_name in subset_attributes:
    true_count = 0
    false_count = 0
    missing_count = 0
    
    for sample_id in gut_NCBI_dict.keys():
        attrs = gut_NCBI_dict[sample_id]['Attributes']
        if attr_name in attrs:
            value = attrs[attr_name]
            if value.lower() == 'true':
                true_count += 1
            elif value.lower() == 'false':
                false_count += 1
            else:
                missing_count += 1
        else:
            missing_count += 1
    
    subset_counts[attr_name] = {
        'true': true_count,
        'false': false_count,
        'missing': missing_count,
        'total': true_count + false_count + missing_count
    }

# Display results
print("=" * 110)
print("SUBSET ATTRIBUTES: TRUE vs FALSE distribution across all samples")
print("=" * 110)
print(f"{'Attribute':<35} {'TRUE':<15} {'FALSE':<16} {'Missing':<17} {'Total':<10}")
print("-" * 110)

for attr_name, counts in subset_counts.items():
    true_pct = (counts['true'] / counts['total'] * 100) if counts['total'] > 0 else 0
    false_pct = (counts['false'] / counts['total'] * 100) if counts['total'] > 0 else 0
    missing_pct = (counts['missing'] / counts['total'] * 100) if counts['total'] > 0 else 0
    
    print(f"{attr_name:<35} {counts['true']:<6} ({true_pct:>5.1f}%)  {counts['false']:<6} ({false_pct:>5.1f}%)  {counts['missing']:<6} ({missing_pct:>5.1f}%)  {counts['total']:<10}")

print("=" * 110)
````
